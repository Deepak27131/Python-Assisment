```python
# selecting all categorical colum ns 
cat_col = df.select_dtypes(include='object').columns
num_col = df.select_dtypes(include='number').columns
```

```python
df[cat_col].isnull().sum()
```

```python
df[num_col].isnull().sum()
```

```python
df.isna().sum()
```

```python
num_col
```

```python
cat_col
```

```python
df['Alley'].unique()
```

```python
df['Alley'].mode()
```

```python
mode = df['Alley'].mode()[0]
df['Alley'].fillna(mode,inplace=True)
```

```python
for col in cat_col:
    mode = df[col].mode()[0]
    df['Alley'].fillna(mode,inplace=True)
    
```

```python

for col in num_col:
    means = df[col].median()
    df[col].fillna(means,inplace=True)
```

```python
df[num_col] = df[num_col].fillna(df[num_col].median())
```
