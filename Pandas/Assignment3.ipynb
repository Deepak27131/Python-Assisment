## Scenario 1: Missing Values

The shipping company has discovered that several passenger records are incomplete. Before
generating reports, management wants to understand the quality of the data.
Tasks:
1. Which columns contain missing information?
2. Which column has the highest number of missing values?
3. Display all passengers whose age is not recorded.
4. Display all passengers whose cabin information is unavailable.
5. If age is mandatory for analysis, how many records would remain after removing missing ages?
6. Suggest a way to retain passengers with missing ages instead of deleting them.

```python
import pandas as pd

df = pd.read_csv('train.csv')
df.head()
```

```python
df.columns[df.isnull().any()]
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().idxmin()
```

```python
df[df['Age'].isnull()]
```

```python
df['Age'].isnull().sum()
```

```python
df.isnull().sum()['Cabin']
```

```python
df.shape[0] - df['Age'].isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Age']
```
