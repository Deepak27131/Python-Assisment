### Group by in columns

```python
import pandas as pd

df = pd.DataFrame({
    'Employee':['A','B','C','D','E','F'],
    'Department':['IT','HR','IT','HR','Sales','IT'],
    'Salary':[50000,45000,60000,55000,70000,65000]
})

df
```

```python
df.groupby('Department')
```

```python
# department wise avg salary
df.groupby('Department')['Salary'].min()
```

```python
df.groupby('Department')['Salary'].max()
```

```python
df.groupby('Department')['Employee'].count()
```

```python
 #group by in mutiple columns

df.groupby('Department')['Salary'].agg(['min','max','median','sum'])
```

### mergei a common columns 

```python
emp = pd.DataFrame({
    'Emp_ID':[101,102,103],
    'Name':['Vinay','Aman','Riya']
})
```

```python
salary = pd.DataFrame({
    'Emp_ID':[101,102,104],
    'Salary':[50000,45000,60000]
})
```

```python
pd.merge(emp,salary, on='Emp_ID')
```

```python
emp.join(salary,lsuffix="emp_id")
```

```python
pd.merge(emp,salary, on='Emp_ID',how='outer')
```

```python
pd.merge(emp,salary, on='Emp_ID',how='inner')
```
