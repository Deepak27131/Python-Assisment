```python
import pandas as pd
df = pd.DataFrame({
    'Area':[2000,3000,4000,500,6000],
    'No of room':[2,3,4,5,6],
    'city':['Delhi', 'mumbai', 'Jaipure', 'Vanarsi', 'Meerut'],
    'class':['first', 'second', 'third', 'fourth', 'fifth'],
    'Survive':['No', 'No', 'Yes', 'Yes', 'Yes']
})
df
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

scaler =MinMaxScaler()
scale_data = scaler.fit_transform(df[['Area','No of room']])
pd.DataFrame(scale_data,columns=scaler.get_feature_names_out())
```

```python
std = StandardScaler()
scale_df = std.fit_transform(df[['Area','No of room']])
new_df = pd.DataFrame(scale_df,columns=std.get_feature_names_out())
```

```python
new_df
```

```python
new_df['Area'].mean()
```

```python
new_df['Area'].std()
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder()
encoder_df = encoder.fit_transform(df[['city']])
```

```python
encoder_df = encoder.transform(df[['city']])   # यह sparse matrix देगा
encoder_df = encoder_df.toarray()   # अब यह dense numpy array बन जाएगा

```

```python
encoder_df
```

```python
data_df = pd.DataFrame(encoder_df, columns=encoder.get_feature_names_out())
data_df
```

```python
from sklearn.preprocessing import OrdinalEncoder
enc = OrdinalEncoder()
enc_df = enc.fit_transform(df[['class']])
df = pd.DataFrame(enc_df , columns=enc.get_feature_names_out())
df
```
