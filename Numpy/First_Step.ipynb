# Vectorization and Broadcasting
* Vectorization

"Vectorization is the process of applying operations on an entire NumPy array at once without explicitly using Python loops, resulting in faster and more memory-efficient computation."


* What is Broadcasting?

Broadcasting ka matlab hai:

Different shapes wale arrays ko compatible bana kar arithmetic operations perform karna.

NumPy automatically smaller array ko larger array ke size ke according expand kar deta hai.

"Automatically expanding smaller arrays so arithmetic operations can be performed between arrays of different but compatible shapes."

```python
a = [1,2,3,4,5]
b = [10,20,30,40,50]

result = []

for i in range(len(a)):
    result.append(a[i] + b[i])

print(result)
```

```python
# NumPy Vectorization
import numpy as np

a = np.array([1,2,3,4,5])
b = np.array([10,20,30,40,50])

result = a + b

print(result)
```

```python
# addition 
arr = np.array([1,2,3])

print(arr + 10)
```

```python
# Multiplication
arr = np.array([1,2,3])

print(arr * 5)
```

```python
# Square
arr = np.array([1,2,3])

print(arr ** 2)
```

```python
arr = np.array([1,2,3])

print(arr + 10)
```

```python
# Row Vector Broadcasting
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([10,20,30])

print(a + b)
```

```python
# Column Broadcasting
a = np.array([[1],
              [2],
              [3]])

b = np.array([10,20,30])

print(a+b)
```
