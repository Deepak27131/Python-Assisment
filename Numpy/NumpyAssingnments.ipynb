### 5. Boolean Masking

. From an array of numbers 1–50, print all numbers greater than 25.

. Print all even numbers using boolean indexing.

. Replace all negative numbers with 0.

. Find numbers divisible by both 3 and 5.

```python
arr = np.arange(1,51)
arr
```

```python
arr = [x for x in arr if x > 25 ]
arr
```

```python
arr = np.arange(1, 51)   # numbers from 1 to 50
print("Original array:", arr)

greater_25 = arr[arr > 25]
print("Numbers greater than 25:", greater_25)
```

```python
arr = np.arange(1, 51)   # numbers from 1 to 50
print("Original array:", arr)

even_arr = arr[arr % 2 == 0]
print("Even array ", even_arr)
```

```python
arr = np.array([10, -5, 7, -3, 0, -8, 15])
arr[arr < 0] = 0
print("After replacing negatives:", arr)
```

```python
arr = np.arange(1, 51)   # Numbers 1–50
divisible = arr[(arr % 3 == 0) & (arr % 5 == 0)]
divisible
```

### 9. Sorting and Searching

38. Sort an array in ascending order.

39. Find the indices of elements greater than 50.

40. Find the position of the maximum and minimum element

```python
arr = np.array([1,5,2,3,6,10,8])
print(np.sort(arr))
```

```python
arr = np.array([10, 20, 55, 60, 30, 75])
indices = np.where(arr > 50)
print("Indices of elements > 50:", indices)
print("Values:", arr[indices])
```

```python
arr = np.array([10, 20, 55, 60, 30, 75])

max_pos = np.argmax(arr)   # index of max element
min_pos = np.argmin(arr)   # index of min element

print("Max element position:", max_pos, "Value:", arr[max_pos])
print("Min element position:", min_pos, "Value:", arr[min_pos])
```
