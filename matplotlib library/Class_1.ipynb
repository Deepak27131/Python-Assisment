```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    'salary':[20000,30000,40000,50000,60000,70000],
    'experience':[1,2,3,2,4,3],
    'years':[2,1,5,7,6,8]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
plt.figure(figsize = (10,6))
plt.plot(df['salary'], df['experience'],marker = 'o')
plt.plot(df['salary'], df['years'] , marker = 'o')

plt.title("Line Plot for the salary vs experience and salary vs years'")
plt.xlabel('Salary')
plt.ylabel('Experience')
plt.show()
```

```python
df = sns.load_dataset('tips')
```

```python
fig , ax = plt.subplots(2,2, figsize=(12,8))
ax[0,0].hist(df['total_bill'])
ax[0,0].set_title(df['total_bill'])

ax[0,1].scatter(df['total_bill'],df['tip'],alpha=0.7)
ax[1,0].bar(df['total_bill'],df['tip'])
ax[1,1].boxplot(df['total_bill'])
plt.tight_layout()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

#1. Top Left: Histogram (Distribution of Total Bill)
ax[0, 0].hist(df['total_bill'], bins=20, edgecolor='black')
ax[0, 0].set_title('Distribution of Total Bill')
ax[0, 0].set_xlabel('Total Bill')

# 2. Top Right: Scatter Plot (Bill vs Tip)
ax[0, 1].scatter(df['total_bill'], df['tip'], alpha=0.7)
ax[0, 1].set_title('Total Bill vs Tip')
ax[0, 1].set_xlabel('Total Bill')
ax[0, 1].set_ylabel('Tip')
plt.tight_layout()
```

```python
import matplotlib.pyplot as plt

x = range(1, 11)
markers = ['o', 's', '^', 'v', 'D', '*', '+', 'x']

for i, marker in enumerate(markers):
    plt.plot(x, [i*2]*10, marker=marker, linestyle='')

plt.title('Different Matplotlib Markers')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```
