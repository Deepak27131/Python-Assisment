```python
import matplotlib.pyplot as plt

import pandas as pd

data = {
    'salary':[2000,3000,4000,5000,650000],
    'experience':[1,2,3,4,5],
    'Year':[2,1,4,3,5]
}

df = pd.DataFrame(data)
df
```

```python


plt.plot(df['salary'], df['experience'])
plt.plot(df['salary'], df['Year'])
```

```python
plt.plot(df['salary'], df['experience'], marker='o')
plt.plot(df['salary'], df['Year'],marker='*')
plt.xlabel('Salary')
plt.ylabel('Experiences')
plt.title("Line plot for Salary vs Experience vs Year")
plt.show()
```
