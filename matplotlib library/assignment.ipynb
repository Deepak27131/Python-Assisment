```python
import pandas as pd

data = {
    "Month":["Jan","Feb","Mar","Apr","May","Jun"],
    "Sales":[50000,55000,60000,58000,65000,70000],
    "Profit":[8000,9000,10000,9500,12000,13000],
    "Customers":[200,220,250,240,280,300],
    "Region":["North","South","East","West","North","South"]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
import matplotlib.pyplot as plt
plt.plot(df['Month'], df['Sales'], marker='o')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title("showing Sales across all months.")
plt.grid()
plt.show()
```

```python
plt.plot(df['Month'],df['Sales'], marker='o')
plt.plot(df['Profit'], df['Sales'], marker='+')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title("showing Sales across all months.")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

# Sales Line
plt.plot(df['Month'],
         df['Sales'],
         marker='o',
         color='blue',
         label='Sales')

# Profit Line
plt.plot(df['Month'],
         df['Profit'],
         marker='+',
         color='red',
         label='Profit')

plt.xlabel("Month")
plt.ylabel("Amount")
plt.title("Sales and Profit Across All Months")

plt.legend()
plt.grid(True)

plt.show()
```

```python
print(df.loc[df['Profit'].idxmax()])
```

```python
highest_profit_month = df.loc[df['Profit'].idxmax(), 'Month']
print(highest_profit_month)
```

```python
df['Gap'] = df['Sales'] - df['Profit']

largest_gap_month = df.loc[df['Gap'].idxmax(), 'Month']

print(largest_gap_month)
```

```python
plt.figure(figsize=(8,5))

bars = plt.bar(df['Month'],
               df['Customers'],
               color='skyblue')

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        str(bar.get_height()),
        ha='center',
        va='bottom'
    )

plt.xlabel("Month")
plt.ylabel("Customers")
plt.title("Customers Across All Months")
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Create Dashboard
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# -----------------------------
# 1. Sales Line Plot
# -----------------------------
axes[0, 0].plot(
    df['Month'],
    df['Sales'],
    marker='o'
)
axes[0, 0].set_title("Sales Line Plot")
axes[0, 0].set_xlabel("Month")
axes[0, 0].set_ylabel("Sales")
axes[0, 0].grid(True)

# -----------------------------
# 2. Sales Bar Plot
# -----------------------------
axes[0, 1].bar(
    df['Month'],
    df['Sales']
)
axes[0, 1].set_title("Sales Bar Plot")
axes[0, 1].set_xlabel("Month")
axes[0, 1].set_ylabel("Sales")

# -----------------------------
# 3. Customers vs Sales Scatter Plot
# -----------------------------
axes[1, 0].scatter(
    df['Customers'],
    df['Sales']
)
axes[1, 0].set_title("Customers vs Sales")
axes[1, 0].set_xlabel("Customers")
axes[1, 0].set_ylabel("Sales")
axes[1, 0].grid(True)

# -----------------------------
# 4. Sales Histogram
# -----------------------------
axes[1, 1].hist(
    df['Sales'],
    bins=5,
    edgecolor='black'
)
axes[1, 1].set_title("Sales Histogram")
axes[1, 1].set_xlabel("Sales")
axes[1, 1].set_ylabel("Frequency")

# Adjust spacing
plt.tight_layout()

# Show Dashboard
plt.show()
```
